```python
%matplotlib qt

import numpy as np
import matplotlib.pyplot as plt
import time
```

```python
plt.ion()
fig, ax = plt.subplots()
x = np.linspace(0, 2*np.pi, 400)
y = np.sin(x)
(line,) = ax.plot(x, y, lw=2)
ax.set_ylim(-1.5, 1.5)
ax.set_title("Sine wave updating")

fig.canvas.draw()
fig.canvas.flush_events()
```

```python
timestamp = time.time()

try:
    phase = 0.0
    while True:
        phase += 0.1
        y = np.sin(x + phase)
        line.set_data(x, y)           # update data
        # ax.relim(); ax.autoscale_view()  # optional if limits change
        fig.canvas.draw()            # redraw
        fig.canvas.flush_events()    # process GUI events
        print(f"Time check: {time.time()-timestamp}")
        timestamp = time.time()
        time.sleep(0.2)             # simulate work
except KeyboardInterrupt:
    print("Stopped")
```

```python
import cv2

cv2.namedWindow("test", cv2.WINDOW_NORMAL)

cam = cv2.VideoCapture(0)

ret, frame = cam.read()
cv2.imshow("test", frame)

# Hold window
while True:
    if cv2.waitKey(1) & 0xFF == ord('s'):
        break

time.sleep(5)

while True:

    ret, frame = cam.read()

    if not ret:
        continue

    cv2.imshow("test", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cv2.destroyAllWindows()
cam.release()
```
